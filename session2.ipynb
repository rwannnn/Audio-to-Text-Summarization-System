```python
import os 

def extract_audio(video_file: str, output_path: str = ".") -> str:
    
    base_name = os.path.basename(video_file).rsplit('.', 1)[0]
    audio_file = os.path.join(output_path, f"{base_name}.wav")
    
    print(f"Extracting audio from {video_file} to {audio_file}")
    
    os.system(f"ffmpeg -i \"{video_file}\" -ar 16000 -ac 1 -c:a pcm_s16le \"{audio_file}\" -y")
    
```

```python
def format_timestamp(seconds: float) -> str:

    hours = int(seconds // 3600)
    minutes = int((seconds % 3600) // 60)
    seconds = int(seconds % 60)
    
    return f"{hours:02d}:{minutes:02d}:{seconds:02d}"
```

```python
import time
import whisper

model = whisper.load_model("small")

def transcribe_with_whisper(audio_file: str):
    results = []
    
    result = model.transcribe(audio_file)
    
    print(result)
    lang = result["language"]
    
    for segment in result["segments"]:
        
        results.append({
            "timestamp": f"{format_timestamp(segment['start'])} - {format_timestamp(segment['end'])}",
            "text": segment["text"].strip(),
        })
    
    return results, lang
```

```python
from langchain_ollama import ChatOllama
from langchain.prompts import PromptTemplate
from langchain.docstore.document import Document

# Connect to Ollama server
model = ChatOllama(
    model="qwen3:0.6b",
    base_url="http://localhost:11434"
)

# Create prompt template
prompt_template = PromptTemplate(
    input_variables=["text"],
    template="""
    You are a smart assistant. Read the following text and provide a concise summary in English:

    Text:
    {text}

    Summary:
    """
)

# Example text
text = """
Artificial Intelligence is transforming industries worldwide. 
It is improving healthcare, finance, transportation, and education. 
However, it also raises ethical concerns and requires proper regulation.
"""

# Wrap text in a Document
doc = Document(page_content=text)

# Format prompt
prompt = prompt_template.format(text=doc.page_content)

# Run model
response = model.invoke(prompt)

print("📌 Summary:")
print(response.content)
```
